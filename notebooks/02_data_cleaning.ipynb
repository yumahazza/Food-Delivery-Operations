```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
DATA_DIR = Path("../data")

df = pd.read_csv(DATA_DIR / "raw" / "Zomato Dataset.csv")

df.head()
```

```python
df.info()
df.describe(include="all").T
```

```python
df.isnull().sum()
```

```python
# Missing value dijadikan frame
missval = (
    df.isna()
    .sum()
    .to_frame("missing_count")
)

missval['missing_pct'] = (missval['missing_count'] / len(df) * 100)

missval.sort_values('missing_pct', ascending=False)
```
